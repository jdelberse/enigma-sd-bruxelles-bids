```python
import os, re
import pandas as pd
```

```python
root = os.getcwd().replace("/code/wrangling","")
```

```python
# Load, add sub-labels, set sub as index
tsv = pd.read_csv(f"{root}/participants.tsv", sep="\t")
tsv.set_index('participant_id', inplace=True)
```

```python
# Move b16, sd9, sd16 to the temp folder
for sub in tsv.index:
    os.mkdir(f"{root}/derivatives/temp/{sub}")
    for ses in ["ses-b16", "ses-sd9", "ses-sd16"]:
        source = f"{root}/{sub}/{ses}"
        dest = f"{root}/derivatives/temp/{sub}/{ses}"
        os.rename(source, dest)
        print(f"{source.replace(root,'')} >> {dest.replace(root,'')}")
```

```python
# Move it back
for sub in tsv.index:
    for ses in ["ses-b16", "ses-sd9", "ses-sd16"]:
        dest = f"{root}/{sub}/{ses}"
        source = f"{root}/derivatives/temp/{sub}/{ses}"
        os.rename(source, dest)
        print(f"{source.replace(root,'')} >> {dest.replace(root,'')}")
```
